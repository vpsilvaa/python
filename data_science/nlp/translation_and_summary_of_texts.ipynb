Entre as várias arquiteturas disponíveis na Hugging Face, algumas das mais utilizadas são:

> BERT (Bidirectional Encoder Representations from Transformers): Focado em entender o contexto bidirecional de uma palavra em uma frase. É amplamente utilizado para tarefas de compreensão de linguagem, como classificação de texto e perguntas e respostas.

> GPT (Generative Pre-trained Transformer): Projetado para tarefas de geração de texto. O GPT é conhecido por sua capacidade de produzir texto que parece ter sido escrito por um humano, sendo usado em chatbots e aplicações de geração de conteúdo.

> T5 (Text-To-Text Transfer Transformer): O T5 trata todas as tarefas de NLP como problemas de transformação de texto. Para a tarefa de sumarização, ele é treinado para transformar um texto longo em uma versão resumida, mantendo o significado e a essência do conteúdo original.

> BART (Bidirectional and Auto-Regressive Transformers): Este modelo combina as características de modelos bidirecionais, como o BERT, com a capacidade de geração de texto de modelos auto-regressivos, como o GPT. BART é eficaz para tarefas de resumo porque pode considerar o contexto completo do texto enquanto gera resumos coerentes.

> mBART (Multilingual BART): Este é um modelo variante do BART que foi treinado em dados de múltiplos idiomas. Ele é capaz de realizar traduções e outras tarefas de NLP em vários idiomas, tornando-o muito útil para aplicações multilíngues.

> MarianMT: Baseado na arquitetura Transformer, MarianMT é um conjunto de modelos de tradução desenvolvidos pela equipe de pesquisa da Microsoft. Estes modelos são altamente eficientes para tradução automática entre muitos pares de idiomas.

Técnicas para superar os desafios de perda de informações, coerência e limitações do tamanho do texto:

1. **Divisão em seções menores**: Uma técnica comum é dividir o texto longo em seções menores e gerar resumos para cada seção individualmente. Esses resumos podem então ser combinados para formar um resumo final coerente.
2. **Modelos hierárquicos**: Modelos hierárquicos tratam a sumarização em múltiplos níveis. Primeiro, eles geram resumos para seções menores do texto e, em seguida, combinam esses resumos para criar um resumo geral. Isso ajuda a manter a coerência e a relevância.
3. **Revisão humana**: Em muitos casos, a revisão humana pode ser necessária para garantir que os resumos de textos longos sejam precisos e úteis. Humanos podem ajustar manualmente o resumo gerado pelo modelo para corrigir possíveis incoerências ou omissões.

### Como os Modelos de Tradução Funcionam

Modelos de tradução são geralmente baseados em arquiteturas de Transformadores, como o T5 (Text-to-Text Transfer Transformer) e o mBART (Multilingual BART). Essas arquiteturas utilizam codificadores e decodificadores para processar e gerar texto. O processo de tradução envolve três etapas principais:

1. **Codificação**: O texto de origem é convertido em uma representação interna pelo codificador.
2. **Atenção**: O modelo utiliza mecanismos de atenção para focar em diferentes partes do texto de origem enquanto gera a tradução.
3. **Decodificação**: O decodificador gera o texto traduzido a partir da representação interna.

```python
def carregar_modelos_trad_resumo(model_resumo, model_trad, ling_origem, ling_final):
    resumidor = pipeline("summarization", model=model_resumo)
    tradutor = pipeline("translation", model=model_trad, src_lang=ling_origem, tgt_lang=ling_final)

    return resumidor, tradutor

def resumir_traduzir_texto_en_pt(text):
  ling_origem = "eng_Latn"
  ling_final = "por_Latn"

  model_resumo = "facebook/bart-large-cnn"
  model_trad = "facebook/nllb-200-distilled-600M"

  resumidor_texto, tradutor_texto = carregar_modelos_trad_resumo(model_resumo, model_trad, ling_origem, ling_final)

  # resumo
  resumo = resumidor_texto(text, max_length=200, min_length=100)
  resumo_texto = resumo[0]['summary_text']

  # tradução
  traducao = tradutor_texto(resumo_texto, max_length=400)
  traducao_resumo_texto = traducao[0]['translation_text']

  return traducao_resumo_texto
```

## Construindo a aplicação

```python
def traduzir_resumir_textos(texto, ling_origem, ling_final):

  nome_modelo_traducao = 'facebook/nllb-200-distilled-600M'
  nome_modelo_resumo = 'facebook/bart-large-cnn'

  idioma_origem = codigos_linguagens[ling_origem]
  idioma_final = codigos_linguagens[ling_final]

  if idioma_origem == "eng_Latn":
    resumidor = pipeline('summarization', model=nome_modelo_resumo)
    resumo = resumidor(texto, max_length=200, min_length=100)[0]['summary_text']
  else:
    tradutor = pipeline('translation', model=nome_modelo_traducao, src_lang=idioma_origem, tgt_lang='eng_Latn')
    texto_traduzido = tradutor(texto, max_length=512)[0]['translation_text']

    resumidor = pipeline('summarization', model=nome_modelo_resumo)
    resumo = resumidor(texto_traduzido, max_length=200, min_length=100)[0]['summary_text']

  tradutor = pipeline('translation', model=nome_modelo_traducao, src_lang='eng_Latn', tgt_lang=idioma_final)
  texto_traduzido = tradutor(texto, max_length=512)[0]['translation_text']

  return texto_traduzido
```

```python
cod_idiomas = list(codigos_linguagens.keys())

inputs = [
    gr.Textbox(lines=5, label='Insira o texto'),
    gr.Dropdown(cod_idiomas, value='English', label='Texto original'),
    gr.Dropdown(cod_idiomas, value='Portuguese', label='Texto final')
]

outputs = gr.Textbox(lines=5, label='Texto final resumido e traduzido')
```

```python
theme = gr.themes.Ocean(
    primary_hue="fuchsia",
    neutral_hue="neutral",
).set(
    body_background_fill='*block_info_text_color',
    body_background_fill_dark='*button_cancel_border_color',
    body_text_color='*color_accent',
    background_fill_primary_dark='*error_icon_color',
    shadow_drop='*button_secondary_shadow_active'
)

with gr.Blocks(theme=theme) as app:

  gr.Markdown('# Tradução e Resumo')
  gr.Markdown('Esta aplicação traduz texto para português e depois o resume. Modelo de tradução: facebook/nllb-200-distilled-600M. Modelo de sumarização: facebook/bart-large-cnn.')

  with gr.Row():
    with gr.Column():
      texto = gr.Textbox(lines=5, label="Texto de entrada")
      origem = gr.Dropdown(cod_idiomas, value='English', label='Texto original')
      destino = gr.Dropdown(cod_idiomas, value='Portuguese', label='Texto final')

    with gr.Column():
      resumo = gr.Textbox(lines=5, label="Texto resumido")

  botao = gr.Button('Gerar resumo')

  botao.click(
      fn = traduzir_resumir_textos,
      inputs = [texto, origem, destino],
      outputs = resumo
)

app.launch(debug=True)
```
